```python
import os
import pkg_resources

# Get a list of installed packages
installed_packages = {pkg.key for pkg in pkg_resources.working_set}

# Extensions to scan (Python scripts and Jupyter notebooks)
file_extensions = (".py", ".ipynb")

# Folder to scan (your project folder)
project_folder = "/workspaces/Job-Resume-Matching"

# Scan all Python-related files for package usage
used_packages = set()
for root, _, files in os.walk(project_folder):
    for file in files:
        if file.endswith(file_extensions):
            with open(os.path.join(root, file), "r", errors="ignore") as f:
                content = f.read()
                for package in installed_packages:
                    if f"import {package}" in content or f"from {package}" in content:
                        used_packages.add(package)

# Find unused packages
unused_packages = installed_packages - used_packages

print("📌 Installed packages but NOT used in your code:")
for package in unused_packages:
    print(package)
```
